### PYTHON DEMO 1

```python
# Write a Python program to check whether a year is a leap year or not

x = int(input("Enter a year:"))
if x%400==0 & x%100==0 & x%4==0:
    print(x, " is a leap year")
else:
    print(x, " is not a leap year")
```

% - modulo <br>
== - is equal to <br>
, - separator for numbers

### PYTHON DEMO 2

```python
# Write a Python function that returns the sum of the squares of first n natural numbers. Use squaresum as the function name.

def squaresum(n):
    x = 0
    while n >= 1:
        m = n**2
        x = x + m
        n = n - 1
    return x

y = int(input("Enter a number: "))
print(squaresum(y))
```

### PYTHON DEMO 3

```python
from math import sin, cos, pi

def polysum (n, s):
    A = (0.25*n*(s**2))/(sin(pi/n)/cos(pi/n))
    P = s*n
    X = round(A + P**2, 4)
    return X

polysum(3, 7)
```

### A. WORD ROTATION PROBLEM

```python
def rotate_word(text):
    x = len(text)
    y = text[1:x:1] + text[0]
    return y

text = str(input("Enter a Word: "))
rotate_word(text)
```

### B.  USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name):
    FN = first_name.lower()
    LN = last_name.lower()
    f = FN.replace(" ", "")
    l = LN.replace(" ", "")
    print(f + "." + l)

firstname = str(input("Enter first name: "))
lastname = str(input("Enter last name: "))
make_username(firstname, lastname)
```

### C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last] + middle + [first]

print(swap_bookends([1, 2, 3, 4, 5, 6]))
print(swap_bookends(["red", "green", "blue"]))
print(swap_bookends([8, 3]))
```

```python
def frame_word(text):
    x = len(text)
    y = text[x-1] + text[0:x:1] + text[x-1]
    return y

n = str(input("Enter a word: "))
frame_word(n)
```

```python
def format_citation(first_name, last_name, year):
    FN = first_name.replace(" " ,"")
    LN = last_name.replace(" ", "")
    last = LN.upper()
    print(last + ", " + FN[0] + ".", "(" + str(year) + ")")

f = str(input("enter first name: "))
l = str(input("enter last name: "))
y = int(input("enter year: "))
format_citation(f, l, y)
```
